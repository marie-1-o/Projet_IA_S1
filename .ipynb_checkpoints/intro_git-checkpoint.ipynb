## 1. Pourquoi utiliser Git ?

Quand plusieurs personnes travaillent sur le même projet, on veut éviter ceci :

```text
projet_final
projet_final_v2
projet_final_v2_bis
projet_final_Michel_OK
projet_final_VRAIMENT_FINAL
projet_final_VRAIMENT_FINAL_correction_michel
```

Git permet de concaténer les changements de chacun dans **un seul projet** et conserver un **historique des modifications**.

## 2. Le vocabulaire minimal

### Le dépôt - `repository` ou `repo`

Un **repo** est un dossier de projet suivi par Git.

Il existe généralement deux copies du dépôt :

```text
┌─────────────────────────┐          ┌─────────────────────────┐
│      VOTRE ORDINATEUR   │          │         GITHUB          │
│                         │          │                         │
│      dépôt LOCAL        │ ───────► │     dépôt DISTANT       │
│                         │   push   │                         │
│                         │ ◄─────── │                         │
└─────────────────────────┘   pull   └─────────────────────────┘
```

- **local** : la version sur votre ordinateur, sur laquelle vous travaillez ;
- **distant** : la version partagée sur GitHub.

### Les 4 actions à retenir

Les actions git sont faites depuis un `terminal` ouvert dans le dossier de votre projet.

Pour comprendre la démarche de "commit", imaginez que vous envoyez un colis contenant votre travail

| Commande | Action | explication |
|---|---|---|
| `git add` | Préparer | J'ajoute des fichiers dans le colis |
| `git commit -am "..."` | Enregistrer | Je ferme le colis et on met une étiquette |
| `git pull` | Récupérer | Je récupère le travail du serveur |
| `git push` | Envoyer | J'envoie mon colis vers le serveur |

Le cycle quotidien sera donc souvent :

**`pull` → travailler → `add` → `commit` → `push`**

## 4. Créer le dépôt distant sur GitHub

Git sait maintenant qui vous êtes, vous allez créer un fichier partagé.

Créez un dépôt GitHub :

1. connectez-vous à GitHub
2. cliquez sur **New repository** (ou nouveau dépôt)

## 6. `git add` - préparer un changement

Ajoutez un nouveau fichier dans votre repo local (vous pouvez créer un README.md ou simplement ajouter le notebook du TD 1 ou celui-ci après l'avoir sauvagardé)
Effectuez à nouveau `git status`, le fichier apparaît normalement en rouge dans la liste `untracked files`, cela signifie que Git l'a vu, mais l'ignore pour l'instant.

Afin de dire à Git que nous souhaitons **suivre les changement** de ce fichier, on utilise `git add`.

`git add` ne crée **pas encore** de sauvegarde, on ne ferme pas le "colis".

On place seulement le fichier dans la zone des changements que vous voulez inclure dans le prochain commit.

On parle parfois de **staging area** ou de **zone d'index**.

```text
fichier modifié  ── git add ──►  changements préparés  ── git commit ──► ajouté à l'historique
```


Le fichier doit maintenant apparaître en vert dans **Changes to be committed** avec la commande `git status`.

vous pouvez faire `git add [nom du fichier]` pour en ajouter un par un, ou `git add .` pour ajouter **tous les fichiers affichés en rouge** par la commande `git status` (attention, ca peut être une mauvaise idée dans un gros projet).

Vérifiez toujours `git status` avant de faire un commit.

## 7. `git commit` - enregistrer une étape du projet

Un **commit** est un instantané des changements préparés, accompagné d'un message.

Un bon message doit expliquer **ce que vous avez fait**.

✅ Bons exemples :

- `Ajout du README`
- `Nettoyage des valeurs manquantes`
- `Correction du calcul de la loss`
- `Test avec la nouvelle base de données`

❌ À éviter :

- `modif`
- `test`
- `ça marche`
- `final final`

Remplacez le message de commit ci-dessous puis éxécutez le bloc pour faire votre premier commit.
(Vous pouvez aussi taper la commande manuellement depuis un terminal)

````python
!git commit -m "nom du commit"
````

🎉 Vous venez de créer votre **premier commit**.

Affichons l'historique :

````python
!git log --oneline --graph
````

Vous devez voir une ligne ressemblant à :

```text
* a1b2c3d Ajout du README
```

Le code court (`a1b2c3d` ici) identifie ce commit.

## 9. S'authentifier avec un Personal Access Token (PAT)

GitHub n'accepte pas votre mot de passe GitHub comme mot de passe pour un `git push` en HTTPS.

Pour cette activité, nous allons utiliser un **Personal Access Token**, ou **PAT**.

Le jeton joue le rôle d'un mot de passe dédié à Git.

### Créer le jeton

Sur GitHub :

1. cliquez sur votre photo de profil → **Settings**
2. ouvrez **Developer settings**
3. ouvrez **Personal access tokens**
4. choisissez de préférence **Fine-grained tokens** → **Generate new token**
5. donnez-lui un nom explicite, par exemple `Projet ML - ordi info 1`
6. choisissez une date d'expiration adaptée à la durée du projet
7. dans **Repository access**, autorisez uniquement le ou les dépôts nécessaires
8. dans les permissions du dépôt, accordez au minimum **Contents: Read and write** afin de pouvoir lire et pousser le contenu
9. générez le token
10. **copiez-le immédiatement** : GitHub ne vous le réaffichera pas forcément ensuite.

### ⚠️ Règle de sécurité

**Ne mettez jamais votre token dans :**

- votre code Python
- ce notebook
- un fichier texte du projet
- un commit Git
- un message envoyé à un camarade

Un token est un **secret personnel**. S'il est exposé, révoquez-le immédiatement dans GitHub et créez-en un nouveau.

## 10. Premier `push`

Cette étape est à effectuer dans le Terminal.

Placez-vous dans le dossier qui contient votre projet dans lequel vous avez déjà fait un **commit**, puis lancez :

```bash
git push origin main 
```

On précise ici `origin main` pour signifier que l'on met les changements sur la branche principale. par la suite, vous pourrez simplement taper `git push`.

Selon votre installation, deux comportements sont possibles.

### Cas A - Git demande un nom d'utilisateur et un mot de passe

Git peut afficher :

```text
Username for 'https://github.com':
Password for 'https://VOTRE_UTILISATEUR@github.com':
```

- **Username** : votre nom d'utilisateur GitHub ;
- **Password** : collez votre **Personal Access Token** - et __pas votre mot de passe GitHub__.

> Quand vous collez le token dans un terminal, il est fréquent que **rien ne s'affiche** à l'écran. C'est normal : tapez ou collez le jeton, puis `Entrée`.

### Cas B - une fenêtre d'authentification s'ouvre

Git Credential Manager peut ouvrir une fenêtre ou un navigateur et proposer une connexion GitHub. Suivez la procédure affichée. Une fois l'authentification enregistrée, vous n'aurez généralement pas à ressaisir vos identifiants à chaque `push`.

Une fois la commande terminée, rechargez la page de votre dépôt GitHub : le ou les fichiers que vous venez d'ajouter et votre commit doivent apparaître.

## 11. Comprendre `push` et `pull`

### `git push`

Envoie **vos commits locaux** vers GitHub.

```text
ordinateur ───────────────► GitHub
             git push
```

### `git pull`

Récupère les commits publiés sur GitHub par les autres membres du groupe (ou vous depuis une autre machine).


```text
ordinateur ◄─────────────── GitHub
             git pull
```


> **Avant de commencer à travailler : pull.**  
> **Après avoir fait un commit : push.**

## 12. La routine recommandée en groupe

Au début de chaque séance :

```bash
git pull
```

Après quelques modifications cohérentes :

```bash
git status
git add fichier1.py fichier2.py
git commit -am "Message clair"
git push
```

Et ainsi de suite.

Parfois, quelqu'un d'autre peut avoir push des changements pendant que vous travaillez. Si c'est la cas, vous pouvez refaire un `pull` après le commit et éventuellement régler le conflit avant de push.

### Pourquoi faire de petits commits ?

Comparez :

❌ Un commit après 5 jours :

```text
Tout le projet terminé
```

✅ Plusieurs petits commits réguliers :

```text
Ajout du chargement du dataset
Nettoyage des valeurs aberrantes
Ajout du modèle de régression
Ajout du calcul de la RMSE
Correction du graphique de prédiction
```

Si une erreur apparaît, il est alors beaucoup plus facile de comprendre **quand** et **pourquoi** elle est apparue (et à cause de qui).

## 13. Quelques règles pour éviter les problèmes

### 1. Toujours regarder `git status`

Avant un `add`, avant un `commit` ou en cas de doute :

```bash
git status
```

C'est votre commande de diagnostic principale.

### 2. Faire un `pull` régulièrement

Si votre camarade a modifié le projet sur GitHub et que vous travaillez sur une ancienne version, Git devra réconcilier les deux historiques.

### 3. Faire des commits petits et cohérents

On peut commit chaque fois qu'on a atteint une version stable (le code compile et s'exécute sans erreurs).

### 4. Ne pas envoyer les gros fichiers de données

Pour le projet de Machine Learning, les datasets peuvent être volumineux.

Ils ne doivent généralement pas être ajoutés au dépôt Git. Utilisez un fichier `.gitignore` pour indiquer à Git ce qu'il doit ignorer.
Ces fichiers seront 

Exemple :

```text
data/
*.csv
__pycache__/
.venv/
```

Adaptez ces règles à votre projet : si un petit CSV constitue volontairement une donnée d'exemple à partager, il peut bien sûr être versionné.

### 5. Ne jamais versionner de secrets

Mots de passe, tokens, clés API et fichiers contenant des identifiants ne doivent jamais être commités.

## Mémo

| Commande | Rôle |
|---|---|
| `git status` | Voir l'état du dépôt et les fichiers modifiés |
| `git add *nom du fichier*` | Préparer un changement |
| `git commit -am "message"` | Enregistrer une étape dans l'historique |
| `git pull --rebase` | Récupérer les mises à jour du dépôt distant |
| `git push` | Envoyer ses commits vers GitHub |
| `git log --oneline --graph` | Afficher l'historique des commits |
| `git remote -v` | Voir l'adresse du dépôt distant |

### Le cycle à retenir

```text
PULL  →  MODIFIER  →  STATUS  →  ADD  →  COMMIT  →  PULL  →  PUSH
```

**Si vous êtes perdu : ne faites pas de commandes au hasard. Commencez par `git status`, lisez le message, puis demandez de l'aide si vous ne comprenez toujorus pas.**
